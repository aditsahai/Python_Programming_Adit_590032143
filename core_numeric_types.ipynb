```python
units_sold = 150
unit_price = 299.0
revenue = units_sold * unit_price
print(f"Revenue: ₹{revenue:,.2f}")

# Example 2: gross margin percentage
cost = 32_000
sales = 50_000
margin_pct = (sales - cost) / sales * 100
print(f"Gross margin: {margin_pct:.1f}%")

# Example 3: complete cartons and loose items
items, carton_size = 137, 12
print("Cartons:", items // carton_size, "Loose items:", items % carton_size)

# Example 4: compound growth
principal, rate, years = 100_000, 0.08, 3
future_value = principal * (1 + rate) ** years
print(f"Future value: ₹{future_value:,.2f}")
```
