Dictionaries 

It is key-value pair datatype.
It is defined by {}.
We can access an element in dictionary using the key assigned.


```python
#indexing of dictionaries 

d={"id":101,'name':"notebook","cost":4000}
print(d["cost"])
d["stock"]=1000
print(d)
print("name" in d)
print("id" in d)
print(len(d))
print(type(d))
```

```python
product = {"id": 101, "name": "Notebook", "price": 499}
print(product["price"])
print(product.get("stock", 0))
print("name" in product)
print(len(product))
print(dict([("region", "North"), ("sales", 2500)]))
```

```python
product = {"id": 101, "price": 499}
product["stock"] = 25
product.update({"price": 549, "category": "Stationery"})
removed = product.pop("stock")
for key, value in product.items(): print(f"{key}: {value}")
print("Removed:", removed)
```

```python
orders={1001:{"region":"North","amount":1250},1002:{"region":"South","amount":1800},1003:{"region":"North","amount":950}}
totals={}
for order in orders.values():
    region=order["region"]
    totals[region]=totals.get(region,0)+order["amount"]
print(totals)
```

```python
words = 'to be or not to be'.split()
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

```python
defaults = {'theme': 'light', 'page_size': 20}
user = {'theme': 'dark'}
settings = defaults | user
print(settings)
```

```python
codes = {'India': 'IN', 'Japan': 'JP'}
by_code = {code: country for country, code in codes.items()}
print(by_code)
```
