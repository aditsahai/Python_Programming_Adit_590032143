```python
#Working with directories and application of file handling:-

from pathlib import Path

Path("scores.txt").write_text("Asha,91\nKabir,84\nMira,88\n", encoding="utf-8")
scores = []
# The context manager closes the resource automatically when this block ends.
with open("scores.txt", encoding="utf-8") as handle:
    for line in handle:
        name, score_text = line.strip().split(",")
        scores.append((name, int(score_text)))
average = sum(score for _, score in scores) / len(scores)
print("Records:", len(scores))
print("Average:", round(average, 2))
```

```python
#Try,Except,else,finally and multiple extensions:-

values = ["18", "unknown", "7"]
for text in values:
    try:
        number = int(text)
    except ValueError:
        print("Not an integer:", text)
    else:
        print("Converted:", number)
    finally:
        print("Checked:", text)
```

```python
#Try,except,else,finally and multiple extensions:-

values = ["18", 24, "unknown"]
for value in values:
    try:
        number = int(value)
    except (ValueError, TypeError):
        print("Cannot convert:", value)
    else:
        print("Converted:", number)
```

```python
#Try,except,else,finally and multiple extensions:-

operations = [("12", 2), ("bad", 2), ("12", 0)]
for text, divisor in operations:
    try:
        result = int(text) / divisor
    except ValueError:
        print("Input is not numeric")
    except ZeroDivisionError:
        print("Divisor cannot be zero")
    else:
        print("Result:", result)
```

```python
#Raise and assert:-

def percentage(part, whole):
    if whole == 0:
        raise ValueError("whole must not be zero")
    return part / whole * 100

try:
    print(percentage(45, 0))
except ValueError as error:
    print("Validation message:", error)
```

```python
#Raise and assert:-

def validate_score(score):
    if not 0 <= score <= 100:
        raise ValueError("score must be between 0 and 100")
    return score

for score in (82, 105):
    try:
        print("Accepted:", validate_score(score))
    except ValueError as error:
        print("Rejected:", error)
```

```python
#Modules,packages and standard modules:-

from pathlib import Path
import importlib.util

Path("geometry.py").write_text("def area(length, width):\n    return length * width\n", encoding="utf-8")
spec = importlib.util.spec_from_file_location("geometry", "geometry.py")
geometry = importlib.util.module_from_spec(spec)
spec.loader.exec_module(geometry)
print("Area:", geometry.area(8, 5))
```
