```python
class Lamp:
    def __init__(self, room):
        self.room = room
        self.on_off = False

    def switchon(self):
        self.on_off =True
        return self.on_off
l = Lamp("Living room")
print(l.switchon())
```

```python
class Student:
    def __init__(self, name, roll_no, marks_12, cs1, cs2, cs3,
                 ac31, ac32, ac33):
        
        self.name = name
        self.roll_no = roll_no
        self.marks_12 = marks_12

        # CS marks
        self.cs1 = cs1
        self.cs2 = cs2
        self.cs3 = cs3

        # Attendance
        self.ac31 = ac31
        self.ac32 = ac32
        self.ac33 = ac33

    # Calculate average marks
    def average_marks(self):
        avg = (self.cs1 + self.cs2 + self.cs3) / 3
        return avg

    # Calculate average attendance
    def average_attendance(self):
        avg = (self.ac31 + self.ac32 + self.ac33) / 3
        return avg

    # Calculate CGPA
    def calculate_cgpa(self):
        cgpa = self.average_marks() / 10
        return cgpa

    # Check placement eligibility
    def eligible_for_placement(self):

        # Check attendance
        if self.average_attendance() < 60:
            return False

        # Check marks
        if self.average_marks() < 60:
            return False

        # Check CGPA
        if self.calculate_cgpa() <= 8:
            return False

        return True


# Taking input from user
name = input("Enter student name: ")
roll_no = int(input("Enter roll number: "))
marks_12 = float(input("Enter 12th grade marks: "))

cs1 = float(input("Enter CS1 marks: "))
cs2 = float(input("Enter CS2 marks: "))
cs3 = float(input("Enter CS3 marks: "))

ac31 = float(input("Enter AC31 attendance: "))
ac32 = float(input("Enter AC32 attendance: "))
ac33 = float(input("Enter AC33 attendance: "))


# Creating Student object
s1 = Student(name, roll_no, marks_12, cs1, cs2, cs3,
             ac31, ac32, ac33)


# Displaying student details
print("\n----- Student Details -----")

print("Student Name:", s1.name)
print("Roll Number:", s1.roll_no)
print("12th Grade Marks:", s1.marks_12)

print("Average CS Marks:", s1.average_marks())
print("Average Attendance:", s1.average_attendance())
print("CGPA:", s1.calculate_cgpa())


# Checking placement eligibility
if s1.eligible_for_placement():
    print("Student is eligible for placement.")
else:
    print("Student is NOT eligible for placement.")
```
