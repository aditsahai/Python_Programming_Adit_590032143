```python
raw_sales = [1200, 950, None, 1400, 1750]
valid = [x for x in raw_sales if x is not None]
average = sum(valid) / len(valid)
print(f"Total: ?{sum(valid):,.2f}")
print(f"Average: ?{average:,.2f}")
print("Above average:", [x for x in valid if x > average])
```

```python
scores = [45, 81, 73, 92, 38]
passing_scaled = [score + 5 for score in scores if score >= 50]
print(passing_scaled)
```

```python
original = [[1], [2]]
shallow = original.copy()
independent = [row.copy() for row in original]
original[0].append(9)
print(shallow, independent)
```

```python
students = [('Asha', 88), ('Kabir', 95), ('Mira', 91)]
print(sorted(students, key=lambda item: item[1], reverse=True))
```

```python
#Using append and extend built in function for lists.

#append will add only one value at the end of the list.
#extend will add a list at the end of the other list.

l=[1,2,3,4]
l.append("Hello")
print(l)

l1=[4,5,6]
l.extend(l1)
print(l)

l2=[7,8,[9,10],11]
l3=[1,2,4]
l3.extend(l2)
print(l3)
```

```python
#using in operator in list 

l=['p','y','t','h','o','n']
print('p' in l)
```

```python
#Indexing and Slicing of a String l[start,stop,step]

l=[1,2,3,4]
print(l[::])
print(l[1:3])
print(l[2:4:2])
print(l[3])
print(l[-3])
print(l[3:1:-1])
print(l[::-1])
```
