Control Statements

there are of two types:-
1) loops
2) if statements 

If statements:-
    a). if statement
    b). if else statement
    c). if elif else statement

a). if statement - checks for the statement to be true. 
b). if else statement - checks for the statement to be true and false.
c). if elif else statement - checks for as many conditions that the user wants.

Loops:-
    If we want to execute a same set of code multiple times we use loops.

there are two types of loops in python:-
      a). for loop
      b). while loop

Range Function:-
   Creates a iterator which we can unpack it one by one.
   range(start,stop,step) is the syntax for using the function.

Break Function:-
    This Function can help us break out of the  when the condition is met.

Continue Function:-
    This Function continues the block of code when the condition is met.

```python
#no of student who scored over 90

count=0
score=[10,20,30,40,90,92,82]
for each in score:
    if each>=90:
        count=count+1
print("total no of student who scored 90 or over is: ",count)
```

```python
# using break function

score=[40,30,50,70,80]
for each in score:
    if each>=90:
        print(each)
        print("student has scored O+")
        break
else:
    print(("no one has scored O+"))
```

```python
transactions= [('North',1200),('South',1800),("North",950)]
total={}
for region,amount in transactions:
    total[region]=total.get(region,0)+amount
for region,amount in sorted(total.items()):
    print(f"{region:<10} ? {amount:>8,.2f}")

```

```python
#Using while loop to code 

while True:
    a=int(input("Enter a Number: "))
    if a <0:
        continue 
    elif a==0:
        break
    else:
        print(a**2)
print("Program Ends")
```
