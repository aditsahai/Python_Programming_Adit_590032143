Tuples:-

It is an ordered , immutable and iterable sequenced data type.
It is defined by ().
We can also unpackage it.
Tuple elements cannot be replaced.
A Tuple is a hetrogeneous data type.

Immutability means that we can access a datatype but cannot change data inserted into it.

Daily life examples:-
    1. coordinate system
    2. vectors 


```python
point = (10, 20)
x, y = point
single = (42,)
record = "E101", "Aarav", 60000
first, *middle, last = (1, 2, 3, 4, 5)
print(point, x, y)
print(single, type(single))
print(record)
print(first, middle, last)
```

```python
from pathlib import Path
path = Path("session07_demo.txt")
with path.open("w", encoding="utf-8") as f: 
    f.write("North,1200\nSouth,1800\n")
with path.open("a", encoding="utf-8") as f:
    f.write("East,1500\n")
with path.open("r", encoding="utf-8") as f:
    lines=[line.strip() for line in f]
print(lines)
path.unlink()
```

```python
text = "Regional sales: ?1,200 — ?????"
encoded = text.encode("utf-8")
decoded = encoded.decode("utf-8")

print(text, type(text))
print(encoded, type(encoded))
print(encoded ==text)
print(decoded == text)
```

```python
rows=["North,1200","South,1800","invalid","East,1500"]
records=[]
for row in rows:
    parts=row.split(",")
    if len(parts)==2 and parts[1].isdigit(): records.append((parts[0],int(parts[1])))
print(records)
print("Total:",sum(amount for _,amount in records))
```
